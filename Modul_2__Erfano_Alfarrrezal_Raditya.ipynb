```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Tugas Analisis 1
Jika mengalikan list biasa Python dengan float ([5000, 7000, 3000] * 0.9), program akan menghasilkan TypeError karena operator perkalian pada list Python tidak mendukung operasi matematika per elemen—pada list, perkalian hanya berfungsi mengulang isi (duplikasi) dan wajib menggunakan bilangan bulat (integer). Sebaliknya, NumPy Array dirancang khusus untuk komputasi numerik dengan fitur broadcasting, sehingga perkalian 0.9 otomatis diteruskan ke setiap elemen di dalam array untuk menghasilkan nilai terdiskon [4500. 6300. 2700. 10800. 4050.].

```python
import pandas as pd
import numpy as np
# IMPORT MODUL FILES GOOGLE COLAB
from google.colab import files

# 1. DATA LOADING & INSPECTION
df = pd.read_csv('/root/dataset_penjualan_kantin.csv')

print("--- 5 Baris Pertama ---")
print(df.head())

print("\n--- Info Dataset ---")
df.info()

print("\n--- Jumlah Missing Value ---")
print(df.isnull().sum())

print(f"\n--- Jumlah Duplikat: {df.duplicated().sum()} ---")

# 2. DATA CLEANING
# A. Hapus Duplikat
df = df.drop_duplicates()

# B. Standarisasi Teks Kategori
df['kategori'] = df['kategori'].str.title()

# C. Bersihkan Kolom harga_satuan (Menghapus 'Rp' dan titik '.')
df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace('Rp', '', regex=False).str.replace('.', '', regex=False)
df['harga_satuan'] = pd.to_numeric(df['harga_satuan'], errors='coerce')

# Mengisi missing value harga_satuan berdasarkan median harga produk yang sama
df['harga_satuan'] = df.groupby('nama_produk')['harga_satuan'].transform(lambda x: x.fillna(x.median()))

# D. Bersihkan Kolom jumlah_terjual (Menghapus 'pcs')
df['jumlah_terjual'] = df['jumlah_terjual'].astype(str).str.replace(' pcs', '', regex=False)
df['jumlah_terjual'] = pd.to_numeric(df['jumlah_terjual'], errors='coerce')

# Mengisi missing value jumlah_terjual dengan median
df['jumlah_terjual'] = df['jumlah_terjual'].fillna(df['jumlah_terjual'].median())

# E. Standarisasi Kolom Tanggal
bulan_map = {'Agustus': 'August'}
for indo, eng in bulan_map.items():
    df['tanggal'] = df['tanggal'].astype(str).str.replace(indo, eng)

df['tanggal'] = pd.to_datetime(df['tanggal'], dayfirst=True, errors='coerce')

# 3. DATA MANIPULATION
# A. Filtering: Menghapus outlier ekstrim (misal jumlah_terjual > 100)
df_filtered = df[df['jumlah_terjual'] <= 100].copy()

# B. Kolom Turunan: Total Penjualan
df_filtered['total_penjualan'] = df_filtered['jumlah_terjual'] * df_filtered['harga_satuan']

# C. Sorting: Mengurutkan dari transaksi terbesar
df_sorted = df_filtered.sort_values(by='total_penjualan', ascending=False)

# D. Groupby / Agregasi per Kategori
ringkasan_kategori = df_sorted.groupby('kategori').agg({
    'jumlah_terjual': 'sum',
    'total_penjualan': 'sum'
}).reset_index()

print("\n--- Ringkasan Penjualan per Kategori ---")
print(ringkasan_kategori)

# 4. SIMPAN DATASET BERSIH & DOWNLOAD KE LAPTOP
df_sorted.to_csv('dataset_bersih.csv', index=False)
print("\nDataset bersih berhasil disimpan ke 'dataset_bersih.csv'")

# Mengunduh otomatis ke folder Downloads laptop
files.download('dataset_bersih.csv')
```
